```python
#IMPORT PACKAGES AND DATA

import newspaper 
import pandas as pd
import numpy as np

news_csv = pd.read_csv("mediacloud_raw.csv")
```

```python
#WEBSCRAPE ARTICLES NOW

text_all = list()
author_all = list()
keywords_all = list()

#number of errors thrown
er = 0

for i in range(len(news_csv)):
    url  = news_csv["url"][i]

    #if article can be downloaded
    try:
        #article.download()
        #article.parse()
        article = newspaper.article(url)
        news_text = article.text
        #only taking first author if multiple authors
        author = article.authors[0]
        
        article.nlp()
        keywords = article.keywords

    #if article cannot be downloaded
    except Exception as e:
        print(str(i) + " failed")
        news_text = None
        author = None
        keywords = None
        
        er = er + 1
    
    if er%50 == 0 :
        print(str(er) + " articles failed until now")
    
    text_all.append(news_text)
    author_all.append(author)
    keywords_all.append(keywords)

news_csv = news_csv.assign(text = text_all,
                           author = author_all,
                           keywords = keywords_all)
```

```python
#SAVE WEBSCRAPED ARTICLES
news_csv.to_csv("newspaper_raw.csv")
```
